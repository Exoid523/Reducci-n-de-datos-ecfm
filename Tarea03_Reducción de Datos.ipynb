# ### Ejemplos de aplicaciones de Machine Learning en Física de Partículas

1. **Reconocimiento de partículas en colisionadores (como el LHC)**  
   Algoritmos de aprendizaje automático se utilizan para clasificar las trayectorias y señales generadas por los detectores, ayudando a identificar qué tipo de partícula ha pasado y con qué energía. Esto reduce la necesidad de intervención humana y mejora la velocidad del análisis.

2. **Filtrado de eventos relevantes en tiempo real**  
   Debido al gran volumen de datos generados por colisiones en experimentos como los del CERN, se usa Machine Learning para seleccionar automáticamente los eventos más interesantes (como posibles señales de partículas nuevas), descartando aquellos que no son útiles para el análisis físico.

3. **Reconstrucción de vértices de decaimiento**  
   Algunos modelos se entrenan para predecir el punto exacto donde una partícula inestable se desintegra, lo cual es esencial para estudios de partículas como los mesones B. El aprendizaje automático mejora la precisión frente a métodos clásicos de reconstrucción.


### Métricas de evaluación para algoritmos de clasificación

Para evaluar el rendimiento de un modelo de clasificación, se utilizan diversas métricas. Estas permiten entender qué tan bien está funcionando el modelo y en qué aspectos podría mejorar. A continuación, se presentan las más comunes:

- **Accuracy (Exactitud)**  
  Mide la proporción de predicciones correctas sobre el total. Es útil cuando las clases están balanceadas, pero puede ser engañosa si hay un desbalance en los datos.

- **Precision (Precisión)**  
  Indica qué proporción de las predicciones positivas fueron realmente correctas. Es importante cuando el costo de falsos positivos es alto.

- **Recall (Sensibilidad o Exhaustividad)**  
  Mide qué proporción de los casos positivos reales fueron correctamente identificados. Es clave en contextos donde los falsos negativos son críticos.

- **F1-Score**  
  Es la media armónica entre precisión y recall. Es útil cuando se busca un equilibrio entre ambos, especialmente con clases desbalanceadas.

- **Matriz de confusión**  
  Es una tabla que muestra los verdaderos positivos, verdaderos negativos, falsos positivos y falsos negativos. Ayuda a analizar en detalle los errores del modelo.

- **AUC-ROC**  
  Representa la capacidad del modelo para distinguir entre clases. Cuanto más cerca esté el valor de 1, mejor será el desempeño.

Estas métricas deben ser elegidas según el contexto del problema, ya que cada una resalta diferentes aspectos del rendimiento del clasificador.


### ¿Qué son los hiperparámetros?

Los **hiperparámetros** son configuraciones externas al modelo que se definen antes del entrenamiento y que afectan directamente su comportamiento. Son claves para controlar la complejidad y el rendimiento del modelo. Algunos ejemplos comunes son:

- **Tasa de aprendizaje (learning rate):** determina cuánto ajusta el modelo los pesos en cada iteración.
- **Número de capas o neuronas en una red neuronal:** controla la arquitectura y capacidad de aprendizaje del modelo.
- **Profundidad de un árbol de decisión:** limita cuántos niveles puede tener el árbol, afectando su complejidad.
- **Número de iteraciones (epochs):** cuántas veces se recorre el conjunto de datos durante el entrenamiento.
- **Parámetros de regularización:** ayudan a evitar el sobreajuste penalizando modelos muy complejos.

A diferencia de los parámetros (que el modelo aprende durante el entrenamiento), los hiperparámetros se deben ajustar manualmente o usando técnicas como **búsqueda aleatoria (Random Search)** o **búsqueda en malla (Grid Search)**.


### ¿Qué es el Train-Test Split en ML?

El **Train-Test Split** es una técnica fundamental en **Machine Learning** para evaluar el rendimiento de un modelo. Consiste en dividir el conjunto de datos original en dos subconjuntos:

- **Conjunto de entrenamiento (Train):** se utiliza para entrenar el modelo, es decir, ajustar los parámetros internos del algoritmo.
- **Conjunto de prueba (Test):** se reserva para evaluar el desempeño del modelo con datos que nunca ha visto.

Generalmente, se utiliza una proporción del **80% para entrenamiento** y **20% para prueba**, aunque puede variar según el caso. Este procedimiento permite:

- Verificar si el modelo se está ajustando correctamente o está sobreajustado.
- Obtener una estimación más realista de su rendimiento en datos nuevos.
- Comparar modelos de manera objetiva usando métricas como precisión, recall, F1, entre otras.

Esta técnica es esencial para asegurar que los modelos aprendan patrones reales y no memoricen el conjunto de entrenamiento.
